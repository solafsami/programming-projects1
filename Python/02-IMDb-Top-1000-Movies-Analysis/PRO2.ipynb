```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# 2
#Load the dataset using Pandas (read_csv or read_excel).
movies_data = pd.read_csv("imdb_top_1000.csv")
#First 5 rows using head()
print("===== First 5 Rows =====")
print(movies_data.head())
print("=================================================")
#Dataset shape (number of rows and columns)
print("===== Dataset Shape =====")
print(movies_data.shape)
print("=================================================")
#Column names
print("===== Column Names =====")
print(movies_data.columns)
# 3
#Select specific columns
selected_columns = movies_data[['Series_Title', 'Released_Year','Genre','No_of_Votes','Gross']]
print("Select specific columns:  ",selected_columns)
print("=============================================================")
#Select specific rows
selected_rows = movies_data.iloc[0:5]
print("\nSelected Rows:",selected_rows)
print("=============================================================")
#Remove duplicate rows
movies_data.drop_duplicates(inplace=True)
print("After removing duplicates:", len(movies_data))
print("=============================================================")
#Change the index using set_index()
movies_data.set_index('Released_Year', inplace=True)
print(movies_data.head(10))
print("=============================================================")
#Create a new DataFrame from part of the data
New_movies_data = movies_data[['Series_Title', 'Certificate','IMDB_Rating']].iloc[0:60]
print("✔ New DataFrame created from a subset of the data (New_movies_data).")
print(New_movies_data)
print("=============================================================")
#4
#Mean,Minimum,Maximum
mean_rating = movies_data['IMDB_Rating'].mean()
min_rating = movies_data['IMDB_Rating'].min()
max_rating = movies_data['IMDB_Rating'].max()

print("Mean Rating:", mean_rating)
print("Minimum Rating:", min_rating)
print("Maximum Rating:", max_rating)
print("=============================================================")
#Use describe() to display statistics
print("Statistical Summary:")
print(movies_data['IMDB_Rating'].describe())
print("=============================================================")
#Create a NumPy array from a column in the dataset
ratings_array = np.array(movies_data['IMDB_Rating'])
print("NumPy Array (Ratings):", ratings_array[:10])
print("=============================================================")
#Perform a mathematical operation on an array (e.g., addition or multiplication)
ratings_multiplied = ratings_array * 2
print("Ratings after multiplication:", ratings_multiplied[:10])
print("=============================================================")
#Calculate a statistical value using NumPy (e.g., mean or standard deviation)
np_mean = np.mean(ratings_array)
np_std = np.std(ratings_array)

print("NumPy Mean:", np_mean)
print("NumPy Standard Deviation:", np_std)
print("=============================================================")
#5
#Save the processed data into a new file:CSV or Excel

New_movies_data.to_csv("processed_data.csv", index=True)
#6
movies_data_reset = movies_data.reset_index()
titles = movies_data_reset['Series_Title'].head(5)
ratings = movies_data_reset['IMDB_Rating'].head(5)
votes = movies_data_reset['No_of_Votes'].head(5)
#One must be a Bar Chart

#Bar chart:
#This bar chart displays the IMDB ratings of the top 5 movies.
#Each bar represents a movie and its rating value.
#It is useful for comparing ratings between movies.
#From the chart, we can identify the movies with the highest and lowest ratings.

plt.figure()
plt.bar(titles, ratings)
plt.title("IMDB Ratings (Bar Chart)")
plt.xlabel("Movie Title")
plt.ylabel("Rating")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
#One must be one of the following: Line chart, Scatter plot, Pie chart
#Each chart must include: Title, Axis labels (if applicable), Grid

#Pie Chart:
#This pie chart displays the distribution of ratings among the selected movies.
#Each slice represents a movie’s contribution to the total ratings.
#It is useful for visualizing the proportion of ratings for each movie.
#From the chart, we can identify which movie has the largest share.

plt.figure()
plt.pie(ratings, labels=titles, autopct='%1.1f%%')
plt.title("Ratings Distribution (Pie Chart)")
plt.show()
#Display two charts in one figure using subplots

#Line Chart:
#This line chart displays the number of votes for each movie.
#It connects the data points to show the trend of votes between movies.
#It helps identify which movies received more votes.
#From the chart, we can determine the most popular movies based on voting counts.

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.plot(titles, votes, marker='o',linestyle='--')
plt.title("Votes (Line Chart)")
plt.xlabel("Movie Title")
plt.ylabel("Votes")
plt.xticks(rotation=45)
plt.grid(True)

# Scatter Plot:
#This scatter plot displays the relationship between movie ratings and the number of votes received.
#Each point represents a movie.
#It is useful for identifying whether movies with higher ratings also received more votes.
#From the chart, we can observe if there is a positive relationship between ratings and popularity.

plt.subplot(1,2,2)
plt.scatter(ratings, votes)
plt.title("Rating vs Votes (Scatter)")
plt.xlabel("Rating")
plt.ylabel("Votes")
plt.grid(True)

plt.tight_layout()
plt.show()
```
